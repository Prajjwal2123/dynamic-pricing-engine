```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import sys
sys.path.append("../src")
from data_loader import generate_synthetic_data, basic_data_quality_report

sns.set_theme(style="whitegrid")
%matplotlib inline
```

```python
df = generate_synthetic_data(n_rows=10000)
basic_data_quality_report(df)
```

```python
cols = ["base_price", "competitor_price", "inventory", "demand"]

fig, axes = plt.subplots(2, 2, figsize=(12, 8))
for ax, col in zip(axes.flatten(), cols):
    sns.histplot(df[col], kde=True, ax=ax, color="steelblue")
    ax.set_title(f"Distribution: {col}")
plt.tight_layout()
plt.show()
```

```python
sample = df.sample(1000, random_state=42)

plt.figure(figsize=(8, 5))
sns.scatterplot(data=sample, x="base_price", y="demand",
                hue="category", alpha=0.6, palette="tab10")
plt.title("Price vs Demand by Category")
plt.xlabel("Base Price (₹)")
plt.ylabel("Demand")
plt.legend(bbox_to_anchor=(1.01, 1), loc="upper left")
plt.tight_layout()
plt.show()
```

```python
df["price_gap"] = df["base_price"] - df["competitor_price"]

plt.figure(figsize=(8, 4))
sns.histplot(df["price_gap"], kde=True, color="coral", bins=50)
plt.axvline(0, color="black", linestyle="--", label="Same as competitor")
plt.title("Price Gap: Our Price vs Competitor Price")
plt.xlabel("Price Gap (₹)  [Positive = We're more expensive]")
plt.legend()
plt.show()

print(f"We are cheaper than competitors: {(df['price_gap'] < 0).sum()} times")
print(f"We are more expensive:           {(df['price_gap'] > 0).sum()} times")
```

```python
day_labels = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"]

plt.figure(figsize=(8, 4))
sns.boxplot(data=df, x="day_of_week", y="demand", palette="Blues")
plt.xticks(ticks=range(7), labels=day_labels)
plt.title("Demand Distribution by Day of Week")
plt.xlabel("Day")
plt.ylabel("Demand")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
sns.scatterplot(data=sample, x="inventory_ratio", y="demand", alpha=0.5, color="purple")
sns.regplot(data=sample, x="inventory_ratio", y="demand",
            scatter=False, color="red", label="Trend")
plt.title("Inventory Ratio vs Demand (Scarcity Effect)")
plt.xlabel("Inventory Ratio (0 = empty, 1 = full)")
plt.ylabel("Demand")
plt.legend()
plt.show()
```

```python
numeric_cols = ["base_price", "competitor_price", "inventory_ratio",
                "is_weekend", "is_holiday", "demand",
                "historical_sales_7d", "historical_sales_30d"]

plt.figure(figsize=(10, 7))
sns.heatmap(df[numeric_cols].corr().round(2),
            annot=True, cmap="coolwarm", center=0,
            linewidths=0.5)
plt.title("Feature Correlation Heatmap")
plt.tight_layout()
plt.show()
```

```python
df.to_csv("../data/raw/synthetic_pricing_data.csv", index=False)
print("Dataset saved to data/raw/synthetic_pricing_data.csv")
print(f"Final shape: {df.shape}")
```
